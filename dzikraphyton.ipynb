```python
import pandas as pd
```

```python
#Tabel Data
dataraw = pd.read_excel('dzikraa.xlsx')
dataraw
```

```python
#Tabel Frekuensi
datafrq = pd.crosstab(index=dataraw['Grade'], columns='Frekuensi')
print(datafrq)
```

```python
datafrq.plot(color='#39FF14')
```

```python
#Grafik Batang
datafrq.plot(kind='bar', color='red')
```

```python
warna = ['#00F5D4', '#7B2CBF', '#CCFF00', '#FF007F', '#D97706', '#60A5FA']

#Pie Chart
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.0f%%', legend=False, colors=warna)
```

```python
#Statistika Deskriptif
dataraw["Final Score"] = pd.to_numeric(dataraw["Final Score"], errors='coerce')
dt = dataraw["Final Score"]

#Menghitung Statistika Deskriptif
stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['variance'] = dt.var()
stats['range'] = dt.max() - dt.min()
stats['skewness'] = dt.skew()
stats['kurtosis'] = dt.kurtosis()
print(stats)
```
